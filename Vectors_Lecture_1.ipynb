Multiplication

```python
M = np.array([[2,3,4], [5,7,8], [9,5,6]])
print(M)
```

```python
N = np.array([[3,2],[7,8],[6,5]])
print(N)
```

multiply (or *) → Element‑wise product. Multiplies arrays position‑by‑position; shapes must be identical or broadcastable. No matrix multiplication logic.

matmul (or @) → Matrix product. For 2‑D: classical row‑×‑column multiplication. For N‑D: treats the last two axes as matrices and batches over the leading dimensions (broadcast if needed).

dot → Sum‑over‑axes product. For 2‑D: same as matmul. For N‑D: sums over the last axis of the first array and the second‑to‑last of the second, keeping all other dimensions – no batch behaviour; result shape differs from matmul when both arrays are 3‑D or higher.

```python
Mul = np.dot(M,N)
print(Mul)
```

```python
Mul = np.matmul(M,N)
print(Mul)
```

In machine learning (especially deep learning), the choice is simple:

matmul (@) Use this for nearly everything.

Handles batched linear transformations automatically: (batch, features) @ (features, units) → (batch, units).

Used in fully‑connected layers, attention (Q @ K^T), and any operation where you multiply matrices with a leading batch dimension.

Works cleanly in frameworks like NumPy, PyTorch, TensorFlow, and JAX with the @ operator.

dot — Rarely used in modern ML code.

Mainly for simple 1‑D vector inner products: np.dot(u, v) → scalar (same as u @ v).

Has unpredictable behaviour for 3‑D+ arrays (it does not batch; it sums over non‑matching axes), which leads to subtle bugs.

Therefore, even for vector dot products, many developers prefer @ for consistency.

Rule of thumb: If you see dot in ML code, it’s either legacy NumPy style or a 1‑D dot product. For everything else — including any multi‑dimensional operation reach for matmul/@.

```python
# Two batches of 2x3 matrices
a = np.ones((2, 2, 3))    # shape (2,2,3)
b = np.ones((2, 3, 2))    # shape (2,3,2)

# matmul: batch multiplication, (2,2,3) @ (2,3,2) -> (2,2,2)
c_matmul = np.matmul(a, b)    # or a @ b
print("matmul shape:", c_matmul.shape)   # (2, 2, 2)

# dot: sums over last axis of a (3) and second-to-last of b (3),
#      keeps other axes -> (2,2) from a and (2,2) from b => (2,2,2,2)
c_dot = np.dot(a, b)
print("dot shape:   ", c_dot.shape)      # (2, 2, 2, 2)
```
